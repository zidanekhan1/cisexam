```python
def caesar(text, key):
    min = ord('a')
    max = ord('z')
    size = min + max + 1

    result = ''

    for char in text:
        text_char = ord(char)
        cipher = min + ((text_char - min + key) % size)
        result += chr(cipher)

    return result

def decrypt_caesar(cipher, key):
    min = ord('a')
    max = ord('z')
    size = min + max + 1

    result = ''

    for char in cipher:
        text_char = ord(char)
        cipher = min + ((text_char - min - key) % size)
        result += chr(cipher)

    return result
```

```python
plaintext = input("enter plaintext : ")
key = int(input("enter key : "))
cipher = caesar(plaintext, key)
print("ciphertext : " + cipher)
print("decrypted ciphertext : " + decrypt_caesar(cipher, key))
```

```python
def encrypt_oddeven(text):
    even = ""
    odd = ""

    for i in range(len(text)):
        if i%2==0:
            even += text[i]
        else:
            odd += text[i]

    return even + odd

def decrypt_oddeven(text):
    div = (len(text)+1) // 2
    even = text[:div]
    odd = text[div:]

    result = ""

    for i in range(len(odd)):
        result += even[i]
        result += odd[i]

    if len(even) > len(odd):
        result += even[-1]

    return result

print(encrypt_oddeven("HELLO WORLD"))
print(decrypt_oddeven("HLOWRDEL OL")) 
```

```python
def rail_fence_encrypt(text, key):
    if key <= 1:
        return text

    rails = [''] * key

    rail = 0
    direction = 1

    for char in text:
        rails[rail] += char

        if rail == 0:
            direction = 1
        elif rail == key - 1:
            direction = -1

        rail += direction

    return ''.join(rails)

cipe = rail_fence_encrypt("HELLOWORLD",3)

def rail_fence_decrypt(cipher, key):
    if key <= 1:
        return cipher

    pattern = []
    direction = 1
    rail = 0

    for _ in cipher:
        pattern.append(rail)

        if rail == key - 1:
            direction = -1
        elif rail == 0:
            direction = 1

        rail += direction

    rails = [''] * key
    index = 0

    for r in range(key):
        for p in pattern:
            if p == r:
                rails[r] += cipher[index]
                index += 1

    result = []
    positions = [0]*key

    for r in pattern:
        result.append(rails[r][positions[r]])
        positions[r]+=1

    return "".join(result)

print(rail_fence_decrypt(cipe, 3))
```
